### **Support**

```python
def Altertag (tag):
    if tag.startswith('J'):
        return 'a'
    elif tag.startswith('V'):
        return 'v'
    elif tag.startswith('R'):
        return 'r'
    else:
        return 'n'
```

```python
def TokenizeShow(sentence):
    sent_tok = sent_tokenize(sentence)
    word_tok = word_tokenize(sentence)
    print('The Tokenized Sentences are: ')
    for sent in sent_tok:
        print(sent)
    print('')

    print('The Tokenized Words are: ')
    for i, word in enumerate(word_tok):
        print(f'{word}', end=' ')
        if i != len(word_tok) - 1:
            print('->', end=' ')
    print('')

def StopwordShow(tokens):
    stop_words = set(stopwords.words('english'))
    
    for tok in tokens:
        if tok in stop_words:
            print(tok)
    print()

def StemmingShow(tokens):
    stemmer = SnowballStemmer('english')
    print('The Stemmed Words are:')

    for tok in tokens:
        print(f'{tok:>12} -> {stemmer.stem(tok)}')
        
    print('')

def LemmetizeShow(tokens):
    lemmatizer = WordNetLemmatizer()
    tagged = pos_tag(tokens)
    print('The Lemmetized Words are:')


    for tok, tag in tagged:
        res = lemmatizer.lemmatize(tok, Altertag(tag))
        print(f'{tok:>12} -> {res}')
    print()

def POSShow(tokens):
    print('The POS Words are:')
    tagged = pos_tag(tokens)

    for tok, tag in tagged:
        print(f'{tok:>12} : {tag}')
    print()

def NERShow(tokens):
    print('The NER are drawn below:')
    tagged = pos_tag(tokens)
    chunked = ne_chunk(tagged)
    chunked.draw()
    print()

def FreqDistShow(tokens):
    print('The Frequency Distribution are shown below:')
    freq = FreqDist(tokens)
    for tok, cc in freq.most_common():
        print(f'{tok:>12} : {cc}')
    print()
```

```python
def Processing(sentence):
    tokens = word_tokenize(sentence)

    TokenizeShow(sentence)
    StopwordShow(tokens)

    tokens = [tok.lower() for tok in tokens]
    tokens = [tok for tok in tokens if tok.isalpha()]
    tokens = [tok for tok in tokens if tok not in punctuation]
    tokens = [tok for tok in tokens if tok not in stopwords.words('english')]
    tokens = [SnowballStemmer('english').stem(tok) for tok in tokens]

    POSShow(tokens)
    LemmetizeShow(tokens)
    NERShow(tokens)
    FreqDistShow(tokens)
    tokens = Lemmatize(tokens)

    return tokens

def Lemmatize(tokens):
    res = []
    lemmatizer = WordNetLemmatizer()
    tagged = pos_tag(tokens)

    for tok, tag in tagged:
        temp = lemmatizer.lemmatize(tok, Altertag(tag))
        res.append(temp)

    return res
```

```python
sentence = input('Input Sentence: ')
Processing(sentence)
```
